```python
# imports

import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
openai = OpenAI()
```

```python
detailed_explanation_system_prompt = """
You have the ability to provide detailed explanations to any question thrown at you.
You should answer in the following JSON format.

{
    "answer": [
        {
            "question_category": "This question is related to software engineering",
            "detailed_explanation": "detailed explanation about the question"
        }
    ]
}
"""

def detailed_explanation_user_prompt(question):
    return f"""
    Provide a detailed explanation to the below question:
    {question}
    """

```

```python
# Get gpt-4o-mini to answer, with streaming

def explain(question):
    messages = [
        {"role":"system", "content":detailed_explanation_system_prompt},
        {"role":"user", "content":detailed_explanation_user_prompt(question)}
    ]
    
    stream = openai.chat.completions.create(
        model = MODEL_GPT, 
        messages = messages,
        response_format={"type": "json_object"},
        stream = True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)



# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

explain(question)
```
